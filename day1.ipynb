# Let's make a quick call to a Frontier model to get started, as a preview!

```python

OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

MODEL="qwen3.5:9b  "
```

```python
# To give you a preview -- calling OpenAI with these messages is this easy. Any problems, head over to the Troubleshooting notebook.

message = "Hello! This is my first ever message to you! Hi!"

messages = [{"role": "user", "content": message}]

messages
```

```python

response = ollama.chat.completions.create(model=MODEL, messages=messages)
response.choices[0].message.content
```

```python
# Let's try out this utility YOLO
from ultralytics.models.yolo import YOLO
from pathlib import Path

#model = YOLO("yolo11n.pt")
model = YOLO("yolo26n.pt")  # load a pretrained YOLO26n model


#model = YOLO("yolov8n.pt")


# Run on all maps
map_folder = Path("../assets/Map/Day1")

for img_path in map_folder.rglob("*.png"):
    print(f"Processing: {img_path}")
    
    results = model.predict(
        source=str(img_path),
        imgsz=640,          # or 416 for faster on GTX 1050
        conf=0.25,
        save=True,          # saves result in runs/detect/predict
        project="pubg_maps",
        name=img_path.stem
    )
```

```python

from ultralytics import YOLO

# Load a pretrained YOLO26n model
model = YOLO("yolo26n.pt")

#model = YOLO("yolov8n.pt")

# Train the model on the COCO8 dataset for 100 epochs
train_results = model.train(
    data="../pubg/map/map-seg.yaml",
    epochs=100,
    imgsz=640,
    batch=8,               # recommended for GTX 1050
    device=0,
    workers=2,
    deterministic=False,
)
# Evaluate the model's performance on the validation set
metrics = model.val()

# Perform object detection on an image
results = model("../pubg/assets/Map/Day1/day2m1C4.png")  # Predict on an image
results[0].show()  # Display results

# Export the model to ONNX format for deployment
path = model.export(format="onnx")  # Returns the path to the exported model
```
